```python
import matplotlib.pyplot as plt
import numpy as np

xs = [3.38, 0.35, 2.07, -0.45, -0.56, -1.06, -2.78, 3.08, -3.99, -5.0,
      -3.18, -1.97, 2.37, -1.67, 4.09, -4.60, 2.68, 2.78, -3.79, -1.87]

ys = [15.65, -11.20, -4.00, 18.03, 7.94, 15.32, -4.40, 1.39, -11.92, -90.24,
      6.92, 28.35, -2.41, -5.47, 35.91, -55.53, 0.77, 4.79, -27.05, 2.85]

m, b = 5.68666556, -0.83754722

def ec(x):
    return m * x + b

# Graficar
plt.scatter(xs, ys, color="red", marker="o", label="Datos")

x_line = np.linspace(min(xs)-1, max(xs)+1, 100)
y_line = ec(x_line)
plt.plot(x_line, y_line, color="blue", label=f"y = {m:.4f}x + {b:.2f}")

plt.title("Recta de mínimos cuadrados")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
x = np.array(xs)
y = np.array(ys)
n = len(x)

A = np.array([
    [np.sum(x**6), np.sum(x**5), np.sum(x**4), np.sum(x**3)],
    [np.sum(x**5), np.sum(x**4), np.sum(x**3), np.sum(x**2)],
    [np.sum(x**4), np.sum(x**3), np.sum(x**2), np.sum(x)],
    [np.sum(x**3), np.sum(x**2), np.sum(x), n]
])

B = np.array([
    np.sum((x**3) * y),
    np.sum((x**2) * y),
    np.sum(x * y),
    np.sum(y)
])

a, b, c, d = np.linalg.solve(A, B)

print("a =", a)
print("b =", b)
print("c =", c)
print("d =", d)

print(f"La función cúbica es:")
print(f"y = {a:.4f}x³ + {b:.4f}x² + {c:.4f}x + {d:.4f}")
```

```python
a, b, c, d = 1.044977916824698, 0.03132740575354256, -8.880663970075506, 2.762289919876904

xs0 = np.linspace(-6, 12, 400)

def s_0(x):
    return a * x**3 + b * x**2 + c * x + d

plt.figure(figsize=(15, 6))

for xi, yi in zip(xs, ys):
    plt.scatter([xi], [yi], color='red')

plt.plot(xs0, s_0(xs0), label='s_0(x)', color='blue')

plt.grid(True, linestyle='--', alpha=0.6)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Mínimos cuadrados')
plt.legend()
plt.xlim(-6, 6)
plt.ylim(-100, 50)
plt.show()
```

```python

x = np.array(xs)
y = np.array(ys)
n = len(x)

def ajustar_polinomio(grado):
    coeficientes = np.polyfit(x, y, grado)
    funcion = np.poly1d(coeficientes)
    return coeficientes, funcion

modelos = {
    "Lineal": ajustar_polinomio(1),
    "Cuadrática": ajustar_polinomio(2),
    "Cúbica": ajustar_polinomio(3)
}

for nombre, (coef, funcion) in modelos.items():
    print(f"\nFUNCIÓN {nombre.upper()}")

    if nombre == "Lineal":
        print(f"y = {coef[0]:.6f}x + {coef[1]:.6f}")

    elif nombre == "Cuadrática":
        print(f"y = {coef[0]:.6f}x² + {coef[1]:.6f}x + {coef[2]:.6f}")

    else:
        print(f"y = {coef[0]:.6f}x³ + {coef[1]:.6f}x² + {coef[2]:.6f}x + {coef[3]:.6f}")

resultados = {}

for nombre, (_, funcion) in modelos.items():
    y_aprox = funcion(x)
    error_abs = np.abs(y - y_aprox)

    indice_max = np.argmax(error_abs)
    indice_min = np.argmin(error_abs)

    resultados[nombre] = {
        "error_abs": error_abs,
        "indice_max": indice_max,
        "indice_min": indice_min
    }

    print("\n=================================")
    print(nombre)
    print("=================================")
    print(f"Error absoluto promedio = {np.mean(error_abs):.6f}")
    print(f"Desviación estándar = {np.std(error_abs):.6f}")

    print("\nPunto con error máximo:")
    print(f"x = {x[indice_max]}, y = {y[indice_max]}")
    print(f"Error máximo = {error_abs[indice_max]:.6f}")

    print("\nPunto con error mínimo:")
    print(f"x = {x[indice_min]}, y = {y[indice_min]}")
    print(f"Error mínimo = {error_abs[indice_min]:.6f}")

xs0 = np.linspace(min(x) - 1, max(x) + 1, 500)

colores = {
    "Lineal": "green",
    "Cuadrática": "blue",
    "Cúbica": "yellow"
}

plt.figure(figsize=(15, 6))
plt.scatter(x, y, color="red", s=70, label="Puntos originales")

for nombre, (_, funcion) in modelos.items():
    plt.plot(xs0, funcion(xs0), color=colores[nombre], label=nombre)

    indice_max = resultados[nombre]["indice_max"]
    indice_min = resultados[nombre]["indice_min"]

    plt.scatter(
        x[indice_max],
        y[indice_max],
        color=colores[nombre],
        marker="X",
        s=40,
        label=f"Error máximo {nombre}"
    )

    plt.plot(
        [x[indice_max], x[indice_max]],
        [y[indice_max], funcion(x[indice_max])],
        color=colores[nombre],
        linestyle="--"
    )

    plt.scatter(
        x[indice_min],
        y[indice_min],
        color=colores[nombre],
        marker="D",
        s=40,
        label=f"Error mínimo {nombre}"
    )

plt.grid(True, linestyle="--", alpha=0.6)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Mínimos cuadrados: lineal, cuadrática y cúbica")
plt.legend()
plt.xlim(min(x) - 1, max(x) + 1)
plt.ylim(min(y) - 1, max(y) + 1)
plt.show()
```

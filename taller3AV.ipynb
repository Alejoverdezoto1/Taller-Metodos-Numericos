## Objetivo 1 - graficar la función 
* Vector xs
* Definir función f(x)
* Calcular vector f(xs) = ys

````python
import numpy as np

xs = list(range(-2, 6))
xs = list(np.linspace(-1, 5, 100))
````

````python
def f(x):
    _y = x**3 - 3 * x**2 + x - 1
    print(f"{x:.2f} \t {_y:.2f}")
    return _y


def df(x):
    _dy = 3 * x**2 - 6 * x + 1
    # print(f"{x:.2f} \t {_dy:.2f}")
    return _dy
````

````python
ys = [f(x) for x in xs]
````

````python
import matplotlib.pyplot as plt

plt.plot(xs, ys)
plt.xlabel("xs")
plt.ylabel("ys")
plt.title("Gráfica de ys vs xs")
plt.grid(True)
plt.show()
````

````python

# Define la función y su derivada
def f(x):
    return x**3 - 3*x**2 + x - 1

def df(x):
    return 3*x**2 - 6*x + 1

# Método de Newton desde x0 = -10
xn = []
x = -10

for i in range(50):
    xn.append(x)
    x = x - f(x)/df(x)

# Método de Newton desde x0 = -8
xm = []
v = -8

for i in range(50):
    xm.append(v)
    v = v - f(v)/df(v)

# Graficar iteraciones
plt.figure(figsize=(10,6))

plt.plot(xn, label='x0 = -10', marker='o')
plt.plot(xm, label='x0 = -8', marker='x')

plt.xlabel('Iteración')
plt.ylabel('Valor de x')
plt.title('Convergencia del Método de Newton')
plt.grid(True)
plt.legend()

plt.show()
````

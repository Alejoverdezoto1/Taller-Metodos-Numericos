1.- Crearemos un vector para los tokens

```python
tokens = list (range (1_000_000))
```

```python
print (tokens [:10])
print (tokens [-5:])
```

2. Crearemos un vector con los precios

```python
precio = []
for t in tokens:
    if t <= 272_000:
        precio.append ((8.16 / 272_000) * t) # Funcion mal implementada, necesita correccion
    else:
        costo = 8.16 + ((45 - 8.16) / (1_000_000 - 272_000)) * (t - 272_000)
        precio.append (costo)
```

3. Se grafica apartir de los vectores

```python
import matplotlib.pyplot as plt

plt.plot (tokens, precio)
plt.xlabel ('Tokens')
plt.ylabel ('Precio')
plt.title ('Precio de los tokens')
# plt.xlim(250_000, 350_000)
plt.show ()
```
